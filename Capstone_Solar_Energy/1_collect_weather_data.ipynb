````python
def prepare_outdir(out_path, verbose=False):
    """create (nested) output folders

    Args:
        out_path (string): directory (e.g. 'data', 'data/more_data/')
        verbose (bool, optional): print more information. Defaults to False.
    """
    # get current path (to return to)
    path = os.getcwd()
    if verbose:
        print("Current path:", path)

    # iterate over all directories (if nested)
    my_path = out_path.split("/")
    if verbose:
        print("Output directory",my_path)
    curr_dir = "."
    while len(my_path) > 0:
        # change if last subdirectory is created
        if len(my_path) == 0 or my_path == [""]:
            break

        # track current (sub)directory
        curr_dir = curr_dir + "/" + my_path[0]
        if not my_path[0] in ("", ".") :
            # create subdirectory, if require
            if not os.path.isdir(my_path[0]):
                print(curr_dir, "...created")
                os.mkdir(my_path[0])
            # change working directory to new subdirectory
            os.chdir(my_path[0])
        
        # remove current subdirectory from list
        if len(my_path) > 1:
            my_path = my_path[1:]
        else:
            break # my_path = []
    
    # go back to original working directory
    os.chdir(path)
````

````python
# https://stackoverflow.com/questions/11023530/python-to-list-http-files-and-directories
def get_file_list_from_html(url, ext=''):
    """get available files from current url

    Args:
        url (str): web link
        ext (str, optional): file extension. Defaults to ''.

    Returns:
        list: files (with required extension)
    """
    # ensure correct url format
    if url.endswith("/"):
        url = url[:-1]
    
    # request web page and parse it
    page = requests.get(url).text
    soup = BeautifulSoup(page, 'html.parser')

    # create file list
    file_list = [url + '/' + node.get('href') for node in soup.find_all('a') if node.get('href').endswith(ext)]
    return file_list

````

````python
location_dict = {
    "Leuchtturm,Alte,Weser": "Leuchtturm Alte Weser",
    "Sankt,Peter": "Sankt Peter",
    "Lippspringe,,Bad": "Bad Lippspringe",
    "Wahnsdorf,bei,Dresden": "Wahnsdorf bei Dresden",
    "List,auf,Sylt": "List auf Sylt",
    "Au,i.d.Hallertau-Günzenhausen": "Au i.d.Hallertau-Günzenhausen",
    "Bellingen,,Bad": "Bad Bellingen",
    "Bergzabern,,Bad": "Bad Bergzabern",
    "Berleburg,,Bad": "Bad Berleburg",
    "Berleburg,,Bad-Stünzel": "Bad-Stünzel Berleburg",
    "Berlin,Brandenburg": "Berlin Brandenburg",
    "Boll,,Bad (KKst)": "Bad Boll (KKst)",
    "Brückenau,,Bad": "Bad Brückenau",
    "Buchen,,Kr.,Neckar": "Buchen Kr. Neckar",
    "UFS,Deutsche,Bucht": "UFS Deutsche Bucht",
    "Doberan,,Bad-Heiligendamm": "Doberan Bad-Heiligendamm",
    "Driburg,,Bad": "Bad Driburg",
    "Dürrheim,,Bad": "Bad Dürrheim",
    "Elster,,Bad-Sohl": "Bad-Sohl Elster",
    "UFS,TW,Ems": "UFS TW Ems",
    "Selb/Oberfranken (Ernst,Reuter,St": "Selb/Oberfranken (Ernst Reuter St",
    "Gehlberg,Glöckchen,im,Tal": "Gehlberg Glöckchen im Tal",
    "Greifswalder,Oie": "Greifswalder Oie",
    "Groß,Lüsewitz": "Groß Lüsewitz",
    "Großer,Arber": "Großer Arber",
    "Grosser,Falkenstein": "Grosser Falkenstein",
    "Grosser,Inselsberg": "Grosser Inselsberg",
    "Grund,,Bad": "Bad Grund",
    "Grund,,Bad (Erzbergwerk)": "Bad Grund (Erzbergwerk)",
    "Hamburg-Sankt,Pauli": "Hamburg-Sankt Pauli",
    "Harzburg,,Bad": "Bad Harzburg",
    "Herrenalb,,Bad": "Bad Herrenalb",
    "Hersfeld,,Bad": "Bad Hersfeld",
    "Bevern,,Kr.,Holzminden": "Bevern Kr. Holzminden",
    "Kahler,Asten": "Kahler Asten",
    "Kissingen,,Bad": "Bad Kissingen",
    "Kleiner,Feldberg/Taunus": "Kleiner Feldberg/Taunus",
    "Klein,Wanzleben": "Klein Wanzleben",
    "Köln-Botanischer,Garten": "Köln-Botanischer Garten",
    "Kösen,,Bad": "Bad Kösen",
    "Kohlgrub,,Bad (Rosshof)": "Bad Kohlgrub (Rosshof)",
    "Krozingen,,Bad": "Bad Krozingen",
    "Lauchstädt,,Bad": "Bad Lauchstädt",
    "Lauterberg,,Bad (A)": "Bad Lauterberg (A)",
    "Leuchtturm,Kiel": "Leuchtturm Kiel",
    "Malente,,Bad-Gremsmühlen": "Malente Bad-Gremsmühlen",
    "Marienberg,,Bad": "Bad Marienberg",
    "Marner,Deich": "Marner Deich",
    "Meinberg,,Bad": "Bad Meinberg",
    "Münder,,Bad": "Bad Münder",
    "Muskau,,Bad": "Bad Muskau",
    "Neuenahr,,Bad-Ahrweiler": "Bad Neuenahr-Ahrweiler",
    "Neuhaus,am,Rennweg": "Neuhaus am Rennweg",
    "Neuhausen,ob,Eck": "Neuhausen ob Eck",
    "Neuhausen,ob,Eck-Unterschwandorf": "Neuhausen ob Eck-Unterschwandorf",
    "Neukirchen/Oberpfalz,-Haid": "Neukirchen/Oberpfalz-Haid",
    "Neustadt,am,Kulm-Filchendorf": "Neustadt am Kulm-Filchendorf",
    "Rappenau,,Bad": "Bad Rappenau",
    "Rauschberg,bei,Ruhpolding": "Rauschberg bei Ruhpolding",
    "Weingarten,,Kr.,Ravensburg": "Weingarten Kr. Ravensburg",
    "Reichenhall,,Bad": "Bad Reichenhall",
    "Renningen-Ihinger,Hof": "Renningen-Ihinger Hof",
    "Altheim,,Kreis,Biberach": "Altheim Kreis Biberach",
    "Rothenfelde,,Bad": "Bad Rothenfelde",
    "Kreuznach,,Bad": "Bad Kreuznach",
    "Saarbrücken-Sankt,Johann": "Saarbrücken-Sankt Johann",
    "Sachsa,,Bad": "Bad Sachsa",
    "Säckingen,,Bad (Bergseestr.)": "Bad Säckingen (Bergseestr.)",
    "Salzuflen,,Bad": "Bad Salzuflen",
    "Sankt,Blasien": "Sankt Blasien",
    "Sankt,Michaelisdonn": "Sankt Michaelisdonn",
    "Sassendorf,,Bad-Ostinghausen": "Bad Sassendorf-Ostinghausen",
    "Schlüsselfeld-Hohn,am,Berg": "Schlüsselfeld-Hohn am Berg",
    "Schömberg,,Kr.,Calw": "Schömberg Kr. Calw",
    "Schwäbisch,Hall-Bibersfeld": "Schwäbisch Hall-Bibersfeld",
    "Schwaigern-Stetten,am,Heuchelberg": "Schwaigern-Stetten am Heuchelberg",
    "Segeberg,,Bad": "Bad Segeberg",
    "Sooden-Allendorf,,Bad": "Bad Sooden-Allendorf",
    "Timmendorfer,Strand": "Timmendorfer Strand",
    "Tölz,,Bad": "Bad Tölz",
    "Tübingen (Botanischer,Garten)": "Tübingen (Botanischer Garten)",
    "Waldhäuser (National,Park)": "Waldhäuser (National Park)",
    "Wildbad,,Bad-Sommerberg": "Bad Wildbad-Sommerberg",
    "Wildungen,,Bad-Braunau": "Bad Wildungen-Braunau",
    "Wittstock-Rote,Mühle": "Wittstock-Rote Mühle",
    "Kohlgrub,,Bad (A)": "Bad Kohlgrub (A)",
    "Kohlgrub,,Bad": "Bad Kohlgrub",
    "Reit,im,Winkl": "Reit im Winkl",
    "Reit,im,Winkl (Tirolerstr.)": "Reit im Winkl (Tirolerstr.)",
    "Münster,am,Stein,,Bad": "Bad Münster am Stein",
    "Ursensollen (Hohenburger,Str.)": "Ursensollen (Hohenburger Str.)",
    "Markt,Erlbach-Mosbach": "Markt Erlbach-Mosbach",
    "Soden,Bad-Salmünster (Kläranlage)": "Soden Bad-Salmünster (Kläranlage)",
    "Urach,,Bad/Erms": "Bad Urach/Erms",
}


def get_weather_station_IDs(out_path, filename, web_tag, data_tag, verbose=False):
    """identify relevant weather stations (data since at least 2005 up to today)

    Args:
        out_path (str): path to save filtered information on weather stations
        filename (str): description of the data (e.g. ST_Stundenwerte_Beschreibung_Stationen.txt)
        web_tag (str): tag used on website (e.g. ST)
        data_tag (str): tag used locally (e.g. sun_hourly)

    Returns:
        list: IDs of relevant weather stations
    """
    print("Identifying available data from %s\n" % filename)
    # Note: use bytestring for special German characters
    # otherwise error: "utf-8' codec can't decode byte 0xfc in position 144: invalid start byte"
    
    # collect weather stations with current data only
    outfile_name = 'Current_dwd_stations_%s_%s_data.csv' % (web_tag, data_tag)
    outfile = open(out_path+outfile_name, 'wb')
    print(out_path+outfile_name)
    
    # collect relevant information on time, place and ID
    list_start_dates = []
    list_cities = []
    list_ids = []
    count = 0
    with open("./"+out_path+filename, 'rb') as infile:
        for line in infile:
            if verbose:
                # encode string with umlauts (special German characters)
                if "litz" in line.decode('windows-1252'):
                    print("byte string:", line.strip())
                    print("converted:    ", line.decode('windows-1252'))
            
            # replace spaces by commas for output csv and decode string
            text = re.sub(' +', ',', line.strip().decode('windows-1252'))
            # correct for city names containing spaces
            text = text.replace(",(", " (")
            for key, val in location_dict.items():
                if key in text:
                    text = text.replace(key, val)
                    # print(key, text)
            # check for other city names containing spaces
            if len(text.split(",")) > 8:
                print(text)

            # get optional information on city and time window
            if verbose:
                info = text.split(",")[1:3]
                info.extend(text.split(",")[6:8])
            
            # keep header
            elif not text[0].isdigit() and text[:5]!="-----":
                outfile.write((text+"\n").encode('utf-8'))
            # ensure recent data
            if "20230" in text:
                start_date = text.split(",")[1]
                # ensure long data range
                if start_date[:4] <= "2005":
                    count += 1
                    list_ids.append(text.split(",")[0])
                    list_start_dates.append(start_date)
                    list_cities.append(text.split(",")[6])
                    outfile.write((text+"\n").encode('utf-8'))
                else:
                    if verbose:
                        print("too young:  ", "\t".join(info))
            elif text[:5]!="-----":
                if verbose:
                    print("not current:", "\t".join(info))

    # summarize findings
    print("\n%d German weather stations with current hourly sun data remaining." % count)
    print("  Recording started between %s and %s" % (min(list_start_dates), max(list_start_dates)))
    print("  Cities:", ", ".join(sorted(list_cities)))
    print("  IDs", ", ".join(list_ids))
    outfile.close()
    return list_ids

````

````python
def download_and_unpack_weather_data(out_path, base_url, web_tag, data_tag, list_ids, verbose=False, overwrite=True):
    """collect relevant weather stations (data since at least 2005 up to today)

    Args:
        out_path (str): path to save filtered information on weather stations
        base_url (str): url with *.zip data
        web_tag (str): tag used on website (e.g. ST)
        data_tag (str): tag used locally (e.g. sun_hourly)
        list_ids (list): pre-selectedrelevant IDs
        overwrite (bool, optional): overwrite/update old data
        verbose (bool, optional): verbosity. Defaults to False.
    """

    # get current path
    data_path = os.path.abspath('')
    
    # check, if directory is present or create it
    out_dir = out_path + data_tag
    prepare_outdir(out_path=out_dir)

    # download all zip files
    print("Downloading %d data files to %s...\n" % (len(list_ids), out_dir))

    # check already present data files to not download them again
    txt_files = glob.glob(data_path+"/"+out_path+data_tag+"/*.txt")
    txt_ids = []
    for name in txt_files:
        if "produkt" in name.lower() and name.endswith(".txt"):
            txt_ids.append(name.split("_")[-1].split(".")[0])
        elif not name.endswith(".txt"):
            txt_files.remove(name)
    print("%d data files already present: %s\n" % (len(txt_ids), ", ".join(txt_ids)))
    if len(txt_ids) == len(list_ids) and not overwrite:
        if set(txt_ids) == set(list_ids):
            print("No download required and no update made!\n")
            print("Lengths of available data files:")
            !wc -l */*/*/*prod*
            return

    # track number of downloads
    id_track = [[],[]] # downloaded, present
    print("Downloading to", out_dir)
    url_list = get_file_list_from_html(base_url, ext='')
    for id in sorted(list_ids):
        for url in url_list:
            if id in url:
                filename = url.split("/")[-1]
                break
        # filename = "stundenwerte_%s_%s_row.zip" % (web_tag, id)
        # url = base_url + filename
        # skip existing data files (zip file present or unpacked data file)
        if overwrite or (not os.path.exists(out_dir+"/"+filename) and not id in txt_ids):
            webpage = requests.get(url, allow_redirects=True)
            open(out_dir+"/"+filename, 'wb').write(webpage.content)
            id_track[0].append(id)
            # subprocess.run(['wget', url], stdout=False) # not completing the file download!
            # subprocess.run(['curl', '-O'] + [url], stdout=False) # not completing the file download!
        else:
            id_track[1].append(id)
    if len(id_track[0])!=0:
        print("%d files downloaded from %s: \n\t%s\n" % (len(id_track[0]), base_url, ", ".join(id_track[0])))
    # if len(id_track[1])!=0:
    #     print("%d files were already loaded: %s\n" % (len(id_track[1]), ", ".join(id_track[1])))

    # switch to data directory to unpack the data
    if not os.getcwd().replace("/", "").replace("\\", "").endswith(out_dir.replace("/", "").replace("\\", "")):
        os.chdir(out_dir)

    # unpack all zip files in another subdirectory and clean up later
    files_long = glob.glob(os.path.abspath('')+"/*.zip")
    if files_long != []:
        print(data_path, "with %d zip data files to unpack" % (len(files_long)))
        subprocess.call(['unzip', "-o", "-q", '*.zip'], stdout=True)

        # delete unnecessary files
        files_long = glob.glob(os.path.abspath('')+"/*.html")
        print ("Deleting %d html files" % (len(files_long)))
        subprocess.call(['rm','-r'] + files_long)
        files_long = glob.glob(os.path.abspath('')+"*/Metadaten*.txt")
        print ("Deleting %d metadata files" % (len(files_long)))
        subprocess.call(['rm','-r'] + files_long)

        # move zip files to archive
        files_long = glob.glob(os.path.abspath('')+"/*.zip")
        files_long_moved = [x[:x.rfind("/")] + "/archive" for x in files_long]
        if not os.path.isdir('archive'):
            os.mkdir('archive')
            print('Archive created!')

        if verbose:
            print("Archiving", files_long)
            print("To", files_long_moved)
        else:
            print("Archiving %d zip files" % len(files_long))

        subprocess.call(['mv'] + files_long + files_long_moved, stdout=False, stderr=False)

        print("Finished archiving!")

    # get information on available data
    print("\nLengths of available data files:")
    !wc -l *prod*

    # switch back to original path
    os.chdir(data_path)
````
