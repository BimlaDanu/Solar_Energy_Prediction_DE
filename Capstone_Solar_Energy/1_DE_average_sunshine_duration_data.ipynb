```python
# Define the URL of the website
base_url = 'https://opendata.dwd.de/climate_environment/CDC/regional_averages_DE/monthly/'

# Create a folder to save the downloaded files
download_folder = 'sunshine-duration-monthly_new'
os.makedirs(download_folder, exist_ok=True)

# Function to download a file given its URL and file name
def download_file(url, file_name):
    response = requests.get(url)
    if response.status_code == 200:
        with open(file_name, 'wb') as f:
            f.write(response.content)
        print(f"File {file_name} downloaded and saved.")
    else:
        print(f"Failed to download {file_name}. Status code: {response.status_code}")

# Get the webpage content
response = requests.get(base_url)
soup = BeautifulSoup(response.content, 'html.parser')

# Find all links on the webpage
links = soup.find_all('a')

# Filter links that point to files and download them
for link in links:
    href = link.get('href')
    if href.endswith('.txt'):
        file_url = base_url + href
        file_name = os.path.join(download_folder, href)
        print(file_name)
        download_file(file_url, file_name)

#print("All files have been downloaded and saved in the folder:", download_folder)
```

```python
import os
import requests

# Define the URL of the website
base_url = 'https://opendata.dwd.de/climate_environment/CDC/regional_averages_DE/'

# Create a folder to save the downloaded files
download_folder = 'sunshine-duration-monthly'
os.makedirs(download_folder, exist_ok=True)

# List of file names to download
file_names = [
    'regional_averages_sd_01.txt',
    'regional_averages_sd_02.txt',
    'regional_averages_sd_03.txt',
    'regional_averages_sd_04.txt',
    'regional_averages_sd_05.txt',
    'regional_averages_sd_06.txt',
    'regional_averages_sd_07.txt',
    'regional_averages_sd_08.txt',
    'regional_averages_sd_09.txt',
    'regional_averages_sd_10.txt',
    'regional_averages_sd_11.txt',
    'regional_averages_sd_12.txt'
]

# Download and save each file in the folder
for file_name in file_names:
    url = base_url + file_name
    file_path = os.path.join(download_folder, file_name)
    response = requests.get(url)
    
    if response.status_code == 404:
        with open(file_path, 'wb') as f:
            f.write(response.content)
        print(f"File {file_name} downloaded and saved.")
    else:
        print(f"Failed to download {file_name}. Status code: {response.status_code}")

print("All files have been downloaded and saved in the folder:", download_folder)
```
